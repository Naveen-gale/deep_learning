```python
# Numerical operations
import numpy as np

# Data visualization
import matplotlib.pyplot as plt

# TensorFlow
import tensorflow as tf

# Load Fashion-MNIST dataset
from tensorflow.keras.datasets import fashion_mnist

# Build CNN model
from tensorflow.keras.models import Sequential

# CNN Layers
from tensorflow.keras.layers import (
    Conv2D,
    MaxPooling2D,
    Flatten,
    Dense,
    Dropout
)

# Evaluate model
from sklearn.metrics import classification_report, confusion_matrix
```

```python
from tensorflow.keras.datasets import fashion_mnist
```

```python
(X_train, y_train), (X_test, y_test) = fashion_mnist.load_data()
```

```python
print(X_train.shape)
print(y_train.shape)
print(X_test.shape)
print(y_test.shape)
```

```python
class_names = [
    "T-shirt/Top",
    "Trouser",
    "Pullover",
    "Dress",
    "Coat",
    "Sandal",
    "Shirt",
    "Sneaker",
    "Bag",
    "Ankle Boot"
]
```

```python
import matplotlib.pyplot as plt

plt.imshow(X_train[0], cmap="gray")
plt.title(class_names[y_train[0]])
plt.axis("off")
plt.show()
```

```python
perseptron = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(10, activation="softmax")

])
```

```python
perseptron.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
hestory_pre = perseptron.fit(
    X_train,
    y_train,
    epochs=10,
    validation_split=0.2
)
```

```python
ann = Sequential([
    Flatten(input_shape=(28, 28)),
    Dense(128, activation="relu"),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])
```

```python
ann.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
hestory_ann = ann.fit(
    X_train,
    y_train,
    epochs=10,
    validation_split=0.2

)
```

```python
cnn = Sequential([
    Conv2D(32, (3, 3), activation="relu", input_shape=(28, 28, 1)),
    MaxPooling2D((2, 2)),
    Conv2D(64, (3, 3), activation="relu"),
    MaxPooling2D((2, 2)),
    Flatten(),
    Dense(64, activation="relu"),
    Dense(10, activation="softmax")
])
```

```python
cnn.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
hestory_cnn = cnn.fit(
    X_train.reshape(-1, 28, 28, 1),
    y_train,
    epochs=10,
    validation_split=0.2
)
```

### Perceptron Model Training History

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(hestory_pre.history['accuracy'], label='Training Accuracy')
plt.plot(hestory_pre.history['val_accuracy'], label='Validation Accuracy')
plt.title('Perceptron Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(hestory_pre.history['loss'], label='Training Loss')
plt.plot(hestory_pre.history['val_loss'], label='Validation Loss')
plt.title('Perceptron Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

### Artificial Neural Network (ANN) Model Training History

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(hestory_ann.history['accuracy'], label='Training Accuracy')
plt.plot(hestory_ann.history['val_accuracy'], label='Validation Accuracy')
plt.title('ANN Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(hestory_ann.history['loss'], label='Training Loss')
plt.plot(hestory_ann.history['val_loss'], label='Validation Loss')
plt.title('ANN Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

### Convolutional Neural Network (CNN) Model Training History

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(hestory_cnn.history['accuracy'], label='Training Accuracy')
plt.plot(hestory_cnn.history['val_accuracy'], label='Validation Accuracy')
plt.title('CNN Model Accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(hestory_cnn.history['loss'], label='Training Loss')
plt.plot(hestory_cnn.history['val_loss'], label='Validation Loss')
plt.title('CNN Model Loss')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()

plt.tight_layout()
plt.show()
```

### Sample Predictions for Each Model

```python
def plot_prediction(model, image, true_label, model_name, class_names, reshape_for_cnn=False):
    plt.figure(figsize=(6, 6))
    plt.imshow(image, cmap="gray")
    plt.axis("off")

    # Prepare image for prediction
    if reshape_for_cnn:
        img_for_pred = image.reshape(1, 28, 28, 1)
    else:
        img_for_pred = image.reshape(1, 28, 28)

    prediction = model.predict(img_for_pred)
    predicted_class = np.argmax(prediction)

    plt.title(f"{model_name} Prediction:\nActual: {class_names[true_label]}\nPredicted: {class_names[predicted_class]} (Probability: {prediction[0][predicted_class]:.2f})")
    plt.show()

# Choose a random image from the test set for prediction
np.random.seed(42) # for reproducibility
idx = np.random.randint(0, len(X_test))
sample_image = X_test[idx]
sample_true_label = y_test[idx]

# Perceptron Prediction
plot_prediction(perseptron, sample_image, sample_true_label, "Perceptron", class_names)

# ANN Prediction
plot_prediction(ann, sample_image, sample_true_label, "ANN", class_names)

# CNN Prediction
plot_prediction(cnn, sample_image, sample_true_label, "CNN", class_names, reshape_for_cnn=True)
```
