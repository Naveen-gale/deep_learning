```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

df = pd.read_csv("validation.csv")

df.head()
```

```python
df = df[['dialog']]
df.head()
```

```python
import string
import re

def clean_dilog(text):
    text = text.lower()
    text = re.sub('\[.*?\]', '', text)
    text = re.sub('https?://\S+|www\.\S+', '', text)
    text = re.sub('<.*?>+', '', text)
    text = re.sub('[%s]' % re.escape(string.punctuation), '', text)
    text = re.sub('\n', '', text)
    text = re.sub('\w*\d\w*', '', text)
    return text.strip()

```

### Apply the Cleaning Function
Now, let's apply the `clean_dilog` function to the 'dialog' column of our DataFrame to clean the text data.

```python
df['dialog'] = df['dialog'].apply(clean_dilog)
display(df.head())
```

```python
corpus = " ".join(df["dialog"])

print(corpus[:500])
words = corpus.split()

print(words[:20])
```

```python

df.head()
```

```python
vocab = sorted(set(words))

word2idx = {
    word:i
    for i,word in enumerate(vocab)
}

idx2word = {
    i:word
    for word,i in word2idx.items()
}

vocab_size = len(vocab)

print(vocab_size)
```

```python
tokens = [word2idx[w] for w in words]

print(tokens[:20])
```

```python
SEQ_LEN = 10
```

```python
X = []
Y = []

for i in range(len(tokens)-SEQ_LEN):

    X.append(tokens[i:i+SEQ_LEN])

    Y.append(tokens[i+SEQ_LEN])
```

```python
import torch

X = torch.tensor(X)

Y = torch.tensor(Y)

print(X.shape)

print(Y.shape)
```

```python
from torch.utils.data import TensorDataset
from torch.utils.data import DataLoader

dataset = TensorDataset(X,Y)

loader = DataLoader(
    dataset,
    batch_size=64,
    shuffle=True
)
```

```python
import torch
import torch.nn as nn

VOCAB_SIZE = vocab_size
EMBED_DIM = 128

embedding = nn.Embedding(
    num_embeddings=VOCAB_SIZE,
    embedding_dim=EMBED_DIM
)
```

```python
sample = X[:2]

print(sample.shape)

out = embedding(sample)

print(out.shape)
```

```python
import math

class PositionalEncoding(nn.Module):

    def __init__(self,d_model,max_len=5000):

        super().__init__()

        pe=torch.zeros(max_len,d_model)

        position=torch.arange(0,max_len).unsqueeze(1)

        div=torch.exp(
            torch.arange(0,d_model,2)
            *
            (-math.log(10000)/d_model)
        )

        pe[:,0::2]=torch.sin(position*div)

        pe[:,1::2]=torch.cos(position*div)

        pe=pe.unsqueeze(0)

        self.register_buffer("pe",pe)

    def forward(self,x):

        return x+self.pe[:,:x.size(1)]
```

```python
position = PositionalEncoding(128)

x = embedding(sample)

x = position(x)

print(x.shape)
```

```python
class SelfAttention(nn.Module):

    def __init__(self,embed_dim):

        super().__init__()

        self.query=nn.Linear(embed_dim,embed_dim)

        self.key=nn.Linear(embed_dim,embed_dim)

        self.value=nn.Linear(embed_dim,embed_dim)

    def forward(self,x):

        Q=self.query(x)

        K=self.key(x)

        V=self.value(x)

        scores=Q@K.transpose(-2,-1)

        scores=scores/math.sqrt(x.size(-1))

        attention=torch.softmax(scores,dim=-1)

        out=attention@V

        return out
```

```python
attention = SelfAttention(128)

y = attention(x)

print(y.shape)
```

```python
class MultiHeadAttention(nn.Module):

    def __init__(self,embed_dim,num_heads):

        super().__init__()

        self.mha=nn.MultiheadAttention(
            embed_dim,
            num_heads,
            batch_first=True
        )

    def forward(self,x):

        out,_=self.mha(x,x,x)

        return out
```

```python
mha = MultiHeadAttention(128,4)

y = mha(x)

print(y.shape)
```

```python
class FeedForward(nn.Module):

    def __init__(self,embed_dim):

        super().__init__()

        self.net=nn.Sequential(

            nn.Linear(embed_dim,512),

            nn.ReLU(),

            nn.Linear(512,embed_dim)

        )

    def forward(self,x):

        return self.net(x)
```

```python
class TransformerBlock(nn.Module):

    def __init__(self,embed_dim,num_heads):

        super().__init__()

        self.attn=MultiHeadAttention(embed_dim,num_heads)

        self.ff=FeedForward(embed_dim)

        self.norm1=nn.LayerNorm(embed_dim)

        self.norm2=nn.LayerNorm(embed_dim)

    def forward(self,x):

        x=self.norm1(x+self.attn(x))

        x=self.norm2(x+self.ff(x))

        return x
```

```python
block = TransformerBlock(128,4)

y = block(x)

print(y.shape)
```

```python
class TinyGPT(nn.Module):

    def __init__(self,
                 vocab_size,
                 embed_dim,
                 num_heads,
                 num_layers):

        super().__init__()

        self.embedding = nn.Embedding(vocab_size, embed_dim)

        self.position = PositionalEncoding(embed_dim)

        self.blocks = nn.Sequential(
            *[
                TransformerBlock(embed_dim, num_heads)
                for _ in range(num_layers)
            ]
        )

        self.fc = nn.Linear(embed_dim, vocab_size)

    def forward(self, x):

        x = self.embedding(x)

        x = self.position(x)

        x = self.blocks(x)

        x = self.fc(x)

        return x
```

```python
model = TinyGPT(
    vocab_size=vocab_size,
    embed_dim=128,
    num_heads=4,
    num_layers=2
)

print(model)
```

```python
criterion = nn.CrossEntropyLoss()
```

```python
optimizer = torch.optim.Adam(
    model.parameters(),
    lr=0.001
)
```

```python
for epoch in range(10):

    total_loss = 0

    for x_batch, y_batch in loader:

        optimizer.zero_grad()

        output = model(x_batch)

        # Use only the last token's prediction
        output = output[:, -1, :]

        loss = criterion(output, y_batch)

        loss.backward()

        optimizer.step()

        total_loss += loss.item()

    print(f"Epoch {epoch+1}: {total_loss:.4f}")
```

```python
import torch

def predict_next_word(text):

    # Convert input words to IDs
    tokens = []

    for word in text.lower().split():

        if word in word2idx:
            tokens.append(word2idx[word])

    # Keep only the last SEQ_LEN words
    tokens = tokens[-SEQ_LEN:]

    # Pad if needed
    while len(tokens) < SEQ_LEN:
        tokens.insert(0, 0)

    x = torch.tensor([tokens])

    with torch.no_grad():

        output = model(x)

        output = output[:, -1, :]

        predicted = torch.argmax(output, dim=1).item()

    return idx2word[predicted]
```

```python
predict_next_word("good morning")
```

```python
predict_next_word("how are")
```

```python
predict_next_word("thank")
```

```python
def generate_text(text, max_words=20):

    for _ in range(max_words):

        next_word = predict_next_word(text)

        text += " " + next_word

    return text
```
